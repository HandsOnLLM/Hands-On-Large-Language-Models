## Model Configuration

MiniMax-M3.1-Flash is the default; MiniMax-M3 is also available. See the [current model documentation](https://platform.minimax.io/docs/guides/models-intro) and [pricing](https://platform.minimax.io/docs/guides/pricing-paygo).

| Model | Context window | Input modalities | Thinking modes |
|-------|---------------:|------------------|----------------|
| MiniMax-M3.1-Flash | 1000000 | text, image, video | adaptive |
| MiniMax-M3 | 1000000 | text, image, video | adaptive, disabled |

Standard pricing below is in USD per million tokens. MiniMax-M3 pricing depends on input length; priority pricing is 1.5 times its standard rates. Cache-write prices are not specified here.

| Model | Input length | Input | Output | Cache read |
|-------|--------------|------:|-------:|-----------:|
| MiniMax-M3.1-Flash | All supported lengths | 0.15 | 0.60 | 0.015 |
| MiniMax-M3 | Up to 512k | 0.30 | 1.20 | 0.06 |
| MiniMax-M3 | Above 512k | 0.60 | 2.40 | 0.12 |

Flash always uses adaptive thinking and defaults to maximum effort. These examples leave thinking and effort unset, so they use the API defaults: M3 defaults to adaptive thinking in Chat Completions and disabled thinking in Messages. Flash does not support disabled thinking, `none` effort, or `service_tier`.

```python
import json
import os

from anthropic import Anthropic
from openai import OpenAI

# Recommended models; explicit model IDs remain usable via MINIMAX_MODEL.
MINIMAX_MODELS = {
    "MiniMax-M3.1-Flash": {
        "context_window": 1000000,
        "max_output_tokens": 524288,
        "input_modalities": ["text", "image", "video"],
        "thinking": ["adaptive"],
    },
    "MiniMax-M3": {
        "context_window": 1000000,
        "input_modalities": ["text", "image", "video"],
        "thinking": ["adaptive", "disabled"],
    },
}
MINIMAX_ENDPOINTS = json.loads("{\"global_en\":{\"openai_base_url\":\"https://api.minimax.io/v1\",\"anthropic_base_url\":\"https://api.minimax.io/anthropic\",\"docs_root\":\"https://platform.minimax.io/docs\"},\"cn_zh\":{\"openai_base_url\":\"https://api.minimaxi.com/v1\",\"anthropic_base_url\":\"https://api.minimaxi.com/anthropic\",\"docs_root\":\"https://platform.minimaxi.com/docs\"}}")

region_name = os.environ.get("MINIMAX_REGION", "global_en")
model_id = os.environ.get("MINIMAX_MODEL", "MiniMax-M3.1-Flash")

if region_name not in MINIMAX_ENDPOINTS:
    raise ValueError(f"Unsupported MiniMax region: {region_name}")

endpoint = MINIMAX_ENDPOINTS[region_name]
model_config = MINIMAX_MODELS.get(model_id, {})
```

## Chat Completions Interface

The selected region supplies the matching OpenAI-compatible base URL.

```python
openai_client = OpenAI(
    api_key=os.environ["MINIMAX_API_KEY"],
    base_url=endpoint["openai_base_url"],
)

openai_response = openai_client.chat.completions.create(
    model=model_id,
    messages=[
        {"role": "user", "content": "Explain context windows in one sentence."}
    ],
)
print(openai_response.choices[0].message.content)
```

## Messages Interface

The same model and region selection can be used with the Anthropic-compatible base URL. Thinking tokens count toward `max_tokens`; increase the budget if the response stops at the limit before producing text. Print text blocks without modifying the full response, which must be retained for any subsequent conversation turns.

```python
anthropic_client = Anthropic(
    api_key=os.environ["MINIMAX_API_KEY"],
    base_url=endpoint["anthropic_base_url"],
)

anthropic_response = anthropic_client.messages.create(
    model=model_id,
    max_tokens=4096,
    messages=[
        {"role": "user", "content": "Explain context windows in one sentence."}
    ],
)
for block in anthropic_response.content:
    if block.type == "text":
        print(block.text)
```

## Switching Configuration

Set MINIMAX_REGION to global_en or cn_zh and MINIMAX_MODEL to MiniMax-M3.1-Flash or MiniMax-M3, then rerun the configuration and request cells. Explicit saved, legacy, or custom model IDs are passed through unchanged; the API determines their availability. Model metadata is empty for IDs outside the recommended catalog. The endpoint dictionary also exposes the matching documentation root.
